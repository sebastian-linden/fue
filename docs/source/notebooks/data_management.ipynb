# Data Management Workflow

Estimating the expected error of a forecasts relies (in the scope of this package) entirely on comparing past forecasts with the true weather outcome and compute the errors. So, before you can really start predicting uncertainty bounds you need data. Howevever, there is an inherent problem with forecasting data. If you fetch a forecast for Tuesday next week, then that forecast is based on the most recent results of the weather models. As soon as there are more uptodate atmospheric measurements available, these models are re-run, which happens about every 3-6 hours. The outdated forecasts are simply overwritten and no longer accessible (at least not that I know off). In this package we are however interested in all forecasts, outdated and new ones and the only way of conserving old forecasts is to fetch and store them regularly.

The default location where this package stores data is under:

`data/raw/forecasts.csv`

When using an instance of a Data object to read the raw data, then it will first look there. However you can pass your own path:

````python
import os

from fue import Data

data = Data()
root_dir = "../../.."
default_path = os.path.join(root_dir, "data/raw/forecasts.csv")
data.read_raw(path=default_path)
data.raw.head()
````

If you don't have any data yet you will have to spend a considerable amount of time collecting data. You can fetch new forecasts either using `Data().fetch_forecast()` + `Data().combine_and_store_forecasts()` or - and this is the much more convenient solution - use the terminal and type:

```batch
fue download
```

I dedicated an entire document on the CLI (Command Line Interface) of this package. This command does the same as running the following two lines. However, the CLI command doesn't have the option to pass a costom path, so if you want to store your data somewhere else, either modify the `cli.py` file in the package or write your own Python script following the steps shown in this notebook.

````python
new_forecast = data.fetch_forecast()
data.combine_and_store_forecasts(new_forecast)
````

Both of these methods will fetch the newest available forecast from Open-Meteo, make sure, that the data has the right format and store it in the specified location. If you execute these lines multiple times in short succession, then only those forecasts will be stored, that are actually new (weather models only update their forecasts every 3-6 hours). Duplicates are discarded automatically.
